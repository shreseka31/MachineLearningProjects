```python
x= finaldf.drop("Level", axis=1).values
y= finaldf['Level'].values
```

```python
type(y)
```

# Splittling dataset into train and test

```python
from sklearn.model_selection import train_test_split
xtrain, xtest, ytrain, ytest= train_test_split(x,y,test_size=0.2,random_state=1)
```

```python
print(xtrain.shape)
print(xtest.shape)
print(ytrain.shape)
print(ytest.shape)
```

# Feature Scaling

```python
from sklearn.preprocessing import StandardScaler 
sc = StandardScaler()
xtrain=sc.fit_transform(xtrain)
```

```python
xtrain
```

# Random Forest Classification

```python
from sklearn.ensemble import RandomForestClassifier
classifier = RandomForestClassifier(n_estimators=10,criterion="entropy",random_state=0)
classifier.fit(xtrain,ytrain)
```

```python
predictions=classifier.predict(sc.fit_transform(xtest))
print(predictions)
print(ytest)
```

```python
from sklearn import metrics
metrics.confusion_matrix(ytest,predictions)
```

```python
metrics.accuracy_score(ytest,predictions)
```

```python
metrics.recall_score(ytest,predictions,average="weighted")
```

# Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
lclassifier=LogisticRegression(random_state=0)
lclassifier.fit(xtrain,ytrain)
```

```python
predictions=lclassifier.predict(sc.fit_transform(xtest))
print(predictions)
print(ytest)
```

```python
from sklearn import metrics
metrics.confusion_matrix(ytest,predictions)
```

```python
metrics.accuracy_score(ytest,predictions)
```

# Decisive Tree

```python
from sklearn.tree import DecisionTreeClassifier
dclassifier = DecisionTreeClassifier(criterion="entropy",random_state=0)
dclassifier.fit(xtrain,ytrain)
```

```python
predictions=dclassifier.predict(sc.fit_transform(xtest))
print(predictions)
print(ytest)
```

```python
from sklearn import metrics
metrics.confusion_matrix(ytest,predictions)
```
