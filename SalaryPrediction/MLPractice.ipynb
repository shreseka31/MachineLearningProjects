```python
import pandas as pd
import numpy as np
```

# Importing data set

```python
df = pd.read_csv("Salary_Data.csv")
```

```python
x = df["YearsExperience"].to_frame()
y = df["Salary"].to_frame()
print(type(x))
print(type(y))
```

# Splitting data into x and y train and test

```python
from sklearn.model_selection import train_test_split
xtrain, xtest, ytrain, ytest= train_test_split(x,y,test_size=0.2,random_state=0)
```

```python
print(ytest.shape)
print(ytrain.shape)
print(xtest.shape)
print(xtrain.shape)
```

# Training Regression model using training set

```python
from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
regressor.fit(xtrain,ytrain)
```

# Predicting the results

```python
regressor.predict(xtest)
```

```python
ytest
```

# Visualizing training set results

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(xtrain,ytrain, color = "green")
plt.plot(xtrain,regressor.predict(xtrain), color = "red")
plt.title("Salary vs Experience")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.show()
```

# Checking the testing set

```python
plt.scatter(xtest,ytest, color = "green")
plt.plot(xtest,regressor.predict(xtest), color = "red")
plt.title("Salary vs Experience")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.show()
```
